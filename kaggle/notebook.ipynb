# FXFunds Deposit and Withdrawal Guide Bot — Benchmarks

Explore 20 synthetic deposit and withdrawal scenarios across USDT (ERC20, TRC20, TON), bank transfer and card, tracked by the FXFunds Deposit and Withdrawal Guide Bot.

- Website: https://fxfunds.com/

*Sample data is synthetic and for demonstration only.*

```python
import glob
import pandas as pd

# Finds the CSV wherever Kaggle mounts the dataset
path = glob.glob('/kaggle/input/**/deposit_withdrawal_benchmarks.csv', recursive=True)[0]
df = pd.read_csv(path)
df.head()
```

## Final status by request type

```python
df.groupby(['type','final_status']).size().unstack(fill_value=0)
```

## Status paths

```python
df[df['status_path'].notna()][['request_id','type','method','status_path']]
```

## Requests rejected at input (validation errors)

```python
df[df['validation'] != 'valid'][['request_id','type','method','validation']]
```

## Reconciliation results

```python
df['reconciliation'].value_counts()
```

## Amount mismatches

```python
mm = df[df['reconciliation'] == 'amount_mismatch'].copy()
mm['difference'] = mm['provider_amount'] - mm['amount']
mm[['request_id','method','amount','provider_amount','difference']]
```

## Blocked actions (duplicate hash, invalid status change)

```python
df[df['last_note'].fillna('').str.startswith('Blocked')][['request_id','final_status','last_note']]
```
